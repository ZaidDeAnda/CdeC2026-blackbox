## Flow control (conditional statements)

Until now, all our code has been executed in a sequential way, one step after the other. What happens if we wanted to have 2 code cases? A code block if X happens, and another one if Y happens? That's why we have flow control!

We can start with the "if" statement, and we can write it this way

```python
if <condition>:
    sentence
else:
    another_sentence
```

Watch out for the indentation!

Usually, in another code languages you mark the start and final of these statements by using curly brackets {}. 
But in Python we use indentations! You can use the tab key or spaces (any quantity of them, like 7 spaces like a psycho), just keep them consistent
